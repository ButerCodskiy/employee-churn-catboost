##Описание датасета
Age
Возраст сотрудника. Обычно используется для анализа, как возраст влияет на удовлетворённость, карьерный рост или вероятность ухода.

Attrition
Целевая переменная, показывающая, ушёл сотрудник (1) или остался (0). Именно по ней часто строят модели прогнозирования оттока.

BusinessTravel
Категориальный признак, описывающий частоту командировок (например, Travel_Rarely, Travel_Frequently). Часто влияет на удовлетворённость работой и баланс между личной жизнью и работой.

DailyRate
Ежедневная ставка (зарплатная составляющая), которая может быть индикатором уровня ответственности или компенсации.

Department
Отдел, в котором работает сотрудник (например, Sales, Research & Development). Этот признак помогает понять, как разные подразделения компании влияют на мотивацию и риск ухода.

DistanceFromHome
Расстояние от дома до работы. Может влиять на удовлетворённость, стресс и, как следствие, вероятность увольнения.

Education
Уровень образования (обычно выражается в числах, где более высокое число означает более высокий уровень). Может коррелировать с должностными обязанностями и карьерным ростом.

EducationField
Область образования (например, Life Sciences, Medical, Other). Позволяет понять профессиональную направленность сотрудников.

EnvironmentSatisfaction
Уровень удовлетворенности рабочей средой (рейтинг, например, от 1 до 4). Важный показатель корпоративного климата.

Gender
Пол сотрудника. Обычно представлен числовым кодированием (например, 0 и 1) и может быть полезен для изучения гендерных различий в отношении к работе.

HourlyRate
Почасовая оплата труда, которая, как и DailyRate или MonthlyIncome, отражает уровень компенсации.

JobInvolvement
Степень вовлеченности в работу (например, в виде рейтинга). Помогает понять, насколько сотрудник заинтересован в достижении целей компании.

JobLevel
Уровень должности или ранговая позиция сотрудника в компании. Обычно отражает иерархическую структуру и может влиять на возможности карьерного роста.

JobRole
Конкретная роль или должность (например, Sales Executive, Research Scientist). Дает более детальное представление о профессиональных обязанностях сотрудника.

JobSatisfaction
Оценка удовлетворенности работой. Часто измеряется по шкале (например, от 1 до 4) и напрямую связана с мотивацией и риском ухода.

MaritalStatus
Семейное положение (Single, Married и т.д.). Может влиять на предпочтения, связанные с балансом между работой и личной жизнью.

MonthlyIncome
Ежемесячный доход сотрудника. Один из ключевых показателей компенсации.

NumCompaniesWorked
Количество компаний, в которых сотрудник ранее работал. Может служить индикатором стабильности или, наоборот, частой смены работы.

OverTime
Признак, указывающий на наличие переработок (обычно 1 – есть, 0 – нет). Часто связан с уровнем стресса и балансом между работой и личной жизнью.

PercentSalaryHike
Процент повышения зарплаты. Может отражать динамику карьерного роста и удовлетворенность компенсацией.

PerformanceRating
Оценка эффективности работы (обычно по шкале, где более высокое значение означает лучшую оценку).

RelationshipSatisfaction
Уровень удовлетворенности отношениями (возможно, внутри коллектива или с руководством). Важный показатель корпоративного климата.

StockOptionLevel
Уровень опционов на акции компании, который может быть частью компенсационного пакета и влиять на лояльность сотрудника.

TotalWorkingYears
Общее количество лет профессионального опыта. Этот показатель помогает оценить стаж и профессиональную зрелость сотрудника.

TrainingTimesLastYear
Количество тренингов или обучающих мероприятий, в которых участвовал сотрудник за последний год. Может отражать инвестиции компании в развитие персонала.

WorkLifeBalance
Оценка баланса между работой и личной жизнью. Обычно выражается по шкале, где более высокие значения означают лучшее соотношение.

YearsAtCompany
Количество лет, которые сотрудник проработал в текущей компании. Часто используется для анализа лояльности.

YearsSinceLastPromotion
Время, прошедшее с последнего повышения. Может сигнализировать о стагнации или, наоборот, стабильности в карьерном росте.

Tenure_to_Age
Отношение стажа работы к возрасту, которое может служить индикатором относительной лояльности и стабильности сотрудника.

Хоть и виден рост дохода с увеличением уровня образования, тем не менее, он не так высок как могло бы ожидаться. Последнее время это вполне нормальное положение на рынке труда - работадатель все меньше полностью полагается на диплом и смотрит на конкретные навыки и опыт. В современных реалиях, образование более доступно и все чаще приобретение конретных навыков не требует обязательного получения нового уровня образования через официальные учреждения.

```python
heatmap_data = pd.crosstab(df['JobSatisfaction'], df['Attrition'])
plt.figure(figsize=(8, 6))
sns.heatmap(heatmap_data, annot=True, fmt='d', cmap='YlGnBu')
plt.title('Уровень удовлетворенности работой и увольнение сотрудников',
          fontsize=16, fontweight='bold', color='#4B3F72')
plt.xlabel('Увольнение (0 - Не уволился, 1 - Уволился)', fontsize=12, fontweight='bold', color='#6A4C5C')
plt.ylabel('Уровень удовлетворенности работой', fontsize=12, fontweight='bold', color='#6A4C5C')
plt.grid(True, linestyle='--', linewidth=0.5, alpha=0.6, color='#D8D8D8')
sns.despine()
plt.tight_layout()
```

Уровень удовлетворенности значительно влияет на частоту увольнения рабочих. Если на самом низком уровне сооотношение составляло 1 уволевшегося на 3.3 оставшихся, то при самом высоком уровне соотношение стало 1 к 8. Причем можно увидеть скачок соотношения между уровнями удовлетворенности 4 и 3.

```python
sns.set_style("white")
dept_attrition = df.groupby(["Department", "Attrition"]).size().unstack().fillna(0)
dept_attrition_percentage = dept_attrition.div(dept_attrition.sum(axis=1), axis=0) * 100
fig, axes = plt.subplots(1, len(dept_attrition_percentage), figsize=(15, 6))
colors = ["#ff6361", "#58508d"]
for ax, (department, values) in zip(axes, dept_attrition_percentage.iterrows()):
    wedges, texts, autotexts = ax.pie(
        values,
        labels=values.index,
        autopct='%1.1f%%',
        colors=colors,
        startangle=140,
        wedgeprops={'edgecolor': 'white', 'linewidth': 2},
        textprops={'fontsize': 12, 'weight': 'bold'}
    )

ax.set_title(department, fontsize=14, fontweight="bold")
plt.suptitle("Attrition по департаментам", fontsize=18, fontweight="bold")
for ax in axes:
    ax.axis('equal')
fig.legend(wedges, values.index, title="Attrition", loc="center left", bbox_to_anchor=(1, 0.5), fontsize=12)
```

В целом распредление везде схожее, однако во втором департаменте мы видим несколько лучшее соотношение, в целом это может быть связано как с самой деятельностью департамента, так и с особыми условиями работы и пр.

```python
from catboost import CatBoostClassifier

model = CatBoostClassifier( loss_function="CrossEntropy",
    random_seed=63,
    iterations=2500,
    learning_rate=0.01,
    early_stopping_rounds=20)
model.fit(X_tr,
    y_train,
    eval_set=(X_te, y_validation),
    verbose=False,)
```

```python
preds_cat = model.predict(X_te)
probs_cat = model.predict_proba(X_te>0.1)[:, 1]
custom = (probs_cat > 0.011).astype(int)
```

```python
preds_cattr = model.predict(X_tr)
probs_cattr = model.predict_proba(X_tr>0.1)[:, 1]
customtr = (probs_cattr > 0.011).astype(int)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score
```

```python
print("TEST CAT_CUSTOM_COD")
print(f'Доля правильных ответов: {accuracy_score(y_validation, preds_cat):.4f}')
print(f'Полнота: {recall_score(y_validation, custom):.4f}')
print(f'Точность: {precision_score(y_validation, preds_cat):.4f}')
print(f'ROC AUC: {roc_auc_score(y_validation, probs_cat):.4f}')
```

```python
print("TRAIN CAT_CUSTOM_COD")
print(f'Доля правильных ответов: {accuracy_score(y_train, preds_cattr):.4f}')
print(f'Полнота: {recall_score(y_train, customtr):.4f}')
print(f'Точность: {precision_score(y_train, preds_cattr):.4f}')
print(f'ROC AUC: {roc_auc_score(y_train, probs_cattr):.4f}')
```

Загружаем нашу модель и оцениваем результаты с выбранными гиперпараметрами на данных с предобработанными категориальными фичами. Основной интерес представляет полнота (recall) — какую долю уволившихся сотрудников мы смогли выявить. Хотя доля правильных ответов(accuracy) на тесте составляет 88%, она малоинформативна из-за несбалансированности классов. А вот точность (precision) тоже довльно хорошая, 80%.

С точки зрения бизнеса работодателю выгоднее получить список всех потенциально увольняющихся сотрудников, будучи уверенным, что в нем действительно есть все, кто уйдет. Это позволит сфокусировать внимание на таких сотрудниках и принять превентивные меры. Accuracy в данном случае приведена лишь как вспомогательная метрика.

ROC-AUC показывает, что в целом модель хорошо справляется с разделением классов (70% на тесте). Однако у модели есть серьезный недостаток — это сильный разброс, особенно по ROC-AUC. Метрики на трейне значительно лучше, чем на тесте, за исключением полноты, но это обусловлено тем, что мы правильно выбрали порог классификации.

 Возможные направления анализа:

Анализ факторов риска увольнения:
Определение ключевых фич (категориальных и числовых), наиболее сильно влияющих на вероятность увольнения. Это позволит выявить закономерности (например, связь между уровнем удовлетворенности, частотой командировок и увольнениями).

Сегментация сотрудников:
Кластеризация сотрудников по профилям (например, по должностям, отделам, опыту работы) для определения групп с повышенным риском ухода. Это может помочь в целенаправленных программах удержания.

Оценка эффективности HR-мер:
Сопоставление предсказаний модели с реальными увольнениями для оценки эффективности предпринятых мер по удержанию сотрудников. Например, как изменился риск увольнения после внедрения новой мотивационной программы.

Анализ воздействия бизнес-показателей:
Исследование корреляций между увольнениями и ключевыми бизнес-показателями (производительность, затраты на обучение, уровень продаж), чтобы понять влияние кадровой стабильности на результаты компании, если быдут иметься дополнительные данные.

Оптимизация модели и порога классификации:
Анализ ошибок модели (ложные срабатывания и пропуски) для улучшения точности прогнозов и определения оптимального порога, учитывая специфику бизнеса и затраты на ложные тревоги.

Отсутствие необходимости долгого подбора гиперпараметров:
Модель начинает работать эффективно практически "из коробки", что ускоряет экспериментальную стадию и настройку гиперпараметров.

Автоматическая обработка категориальных фич:
CatBoost позволяет работать с категориальными признаками «из коробки», без необходимости вручную применять one-hot или label encoding. Это упрощает предварительную обработку данных и снижает риск потери информации, возникающей при явном кодировании категориальных переменных.

Высокая производительность на табличных данных:
Алгоритм градиентного бустинга, лежащий в основе CatBoost, отлично справляется с данными, где присутствуют сложные взаимосвязи между фичами.

Устойчивость к переобучению:
Встроенные механизмы, такие как ранняя остановка помогают контролировать переобучение. Это критично при работе с данными, в которых соотношение классов часто несбалансировано, а модели нужно обеспечить стабильность предсказаний на новых данных.
